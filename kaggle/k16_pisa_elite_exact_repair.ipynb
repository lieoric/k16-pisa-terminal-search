# K16 Pisa v9 — exact repair of the 54 v8 GPU elites

Kaggle settings: **Accelerator: None**, **Internet: On**.

Add the committed output of `chamine/k16-pisa-gpu-hunter` as an input before running. The notebook must see all 54 `shard-*.json` files.

```python
import glob, os, pathlib, shutil, subprocess, sys

REPO = pathlib.Path('/kaggle/working/k16-pisa-terminal-search')
if REPO.exists():
    shutil.rmtree(REPO)
subprocess.run([
    'git', 'clone', '--depth', '1', '--branch', 'agent/k16-endpoint-endgame',
    'https://github.com/lieoric/k16-pisa-terminal-search.git', str(REPO)
], check=True)
subprocess.run([
    sys.executable, '-m', 'pip', 'install', '-q', '-r', str(REPO / 'requirements.txt')
], check=True)

source_files = glob.glob('/kaggle/input/**/shard-*.json', recursive=True)
print('candidate shard files:', len(source_files))
if len(source_files) < 54:
    raise RuntimeError(
        'The v8 output is not attached. Add chamine/k16-pisa-gpu-hunter '
        'as a notebook input, then Save Version again.'
    )

output = pathlib.Path('/kaggle/working/k16_elite_repair_v9')
cmd = [
    sys.executable, str(REPO / 'scripts' / 'v8_elite_exact_repair.py'),
    '--input-glob', '/kaggle/input/**/shard-*.json',
    '--output-dir', str(output),
    '--parallel', '4',
    '--solver-workers', '1',
    '--fiber-seconds', '600',
    '--shell', '1:2:120',
    '--shell', '3:4:600',
    '--shell', '5:6:1200',
]
print(' '.join(cmd), flush=True)
subprocess.run(cmd, check=True, cwd=REPO)
print('results:', output)
```
