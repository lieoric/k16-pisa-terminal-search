# K16 Pisa v6 true-partition GPU witness hunter

Kaggle settings: **GPU T4 x2**, **Internet on**. The two GPUs concurrently search 64 mutually exclusive structural partitions. A miss is not an UNSAT proof. Results are written under `/kaggle/working/k16_gpu_results_v6`.

```python
import os, shutil, subprocess

REPO = "https://github.com/lieoric/k16-pisa-terminal-search.git"
REF = "agent/witness-hunter"  # change to main after the PR is merged
ROOT = "/kaggle/working/k16-pisa-terminal-search"

if os.path.exists(ROOT):
    shutil.rmtree(ROOT)
subprocess.run(["git", "clone", "--depth", "1", "--branch", REF, REPO, ROOT], check=True)
os.chdir(ROOT)

base = ["python", "kaggle/k16_pisa_gpu_hunter.py", "--seconds-per-shard", "300", "--batch-size", "16384"]
commands = [
    base + ["--shard-start", "0", "--shard-count", "32", "--device", "cuda:0", "--output-dir", "/kaggle/working/k16_gpu_results_v6/d7"],
    base + ["--shard-start", "32", "--shard-count", "32", "--device", "cuda:1", "--output-dir", "/kaggle/working/k16_gpu_results_v6/d6"],
]
processes = [subprocess.Popen(command) for command in commands]
return_codes = [process.wait() for process in processes]
if any(return_codes):
    raise RuntimeError(f"GPU worker failure: {return_codes}")
```
