# K16 Pisa GPU witness hunter

Kaggle settings: **GPU on**, **Internet on**. This searches for one SAT witness; a miss is not an UNSAT proof. Results and the checkpoint are written under `/kaggle/working/k16_gpu_results`.

```python
import os, shutil, subprocess

REPO = "https://github.com/lieoric/k16-pisa-terminal-search.git"
REF = "agent/witness-hunter"  # change to main after the PR is merged
ROOT = "/kaggle/working/k16-pisa-terminal-search"

if os.path.exists(ROOT):
    shutil.rmtree(ROOT)
subprocess.run(["git", "clone", "--depth", "1", "--branch", REF, REPO, ROOT], check=True)
os.chdir(ROOT)

subprocess.run([
    "python", "kaggle/k16_pisa_gpu_hunter.py",
    "--shard-start", "0",
    "--shard-count", "64",
    "--seconds-per-shard", "300",
    "--batch-size", "16384",
    "--output-dir", "/kaggle/working/k16_gpu_results",
], check=True)
```
